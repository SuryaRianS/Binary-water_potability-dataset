Identifikasi dataset:

Jenis Klasifikasi: Binary Classification
- Independent Variables: pH, Hardness, Solids, Chloramines, Sulfate, Conductivity, Organic Carbon, Trihalomethanes, dan Turbidity
- Dependent Variable: Potability
- Jumlah Independent Variables: 9 fitur
- Jumlah Dependent Variables: 1 target
- Jumlah Kategori Target: 2 kategori
- Kategori 0: Air tidak layak diminum
- Kategori 1: Air layak diminum

Tujuan: Mengklasifikasikan kelayakan air minum berdasarkan parameter kualitas air

```python
import pandas as pd

df = pd.read_csv("water_potability.csv")

print("IDENTIFIKASI DATASET WATER POTABILITY")

print("\nInformasi Dataset:")
df.info()

print("\nJumlah Baris dan Kolom:")
print(df.shape)

print("\nNama Kolom:")
print(df.columns.tolist())

# Menentukan independent dan dependent variables
target = "Potability"

X = df.drop(columns=[target])
y = df[target]

print("\nIndependent Variables:")
print(X.columns.tolist())

print("\nDependent Variable:")
print(target)

print("\nJumlah Independent Variables:", X.shape[1])
print("Jumlah Dependent Variables:", 1)

print("\nJumlah Kategori Target:")
print(y.nunique())

print("\nDistribusi Target:")
print(y.value_counts().sort_index())

print("\nIdentifikasi Klasifikasi:")
if y.nunique() == 2:
    print("Binary Classification")
else:
    print("Bukan Binary Classification")
```

Penjelasan variabel:
1. pH: Mengukur tingkat keasaman atau kebasaan air.
2. Hardness: Menunjukkan tingkat kesadahan air akibat kandungan mineral.
3. Solids: Menunjukkan jumlah padatan terlarut dalam air.
4. Chloramines: Menunjukkan kandungan chloramine dalam air.
5. Sulfate: Menunjukkan kandungan sulfat.
6. Conductivity: Mengukur kemampuan air menghantarkan listrik.
7. Organic Carbon: Menunjukkan kandungan karbon organik.
8. Trihalomethanes: Menunjukkan kandungan senyawa trihalomethane.
9. Turbidity: Mengukur tingkat kekeruhan air.
